```python
import pandas as pd
import numpy as np
import matplotlib as plt
import seaborn as sns
```

```python
stolen_water = pd.read_csv('./csvs/consum cu posibil furt de apă.csv', sep=';')
# stolen_water.describe()

stolen_water['Date'] = pd.to_datetime(stolen_water['x']).dt.date
stolen_water['Hour'] = pd.to_datetime(stolen_water['x'], format='%d/%m/%Y %H:%M').dt.hour

stolen_water.info()
```

```python
result = stolen_water[stolen_water['Date'] == pd.to_datetime("2025-03-19").date()]

sns.scatterplot(data=result, x='Hour', y='y', hue='Hour')
```
